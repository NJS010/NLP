```python
import gensim
from gensim.models import Word2Vec, KeyedVectors
```

```python
import gensim.downloader as api

# 66 MB download (50-dimensional vectors)
model = api.load('glove-wiki-gigaword-50')
```

```python
print(model['king'].shape)  # (50,)

# Vector arithmetic: king - man + woman = queen
print(model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5))

# Similarity
print(model.similarity('movie', 'film'))
```

```python
vec = model['king'] - model['man'] + model['woman']
model.most_similar([vec])
```

```python
# Analogy: rupee - india + england = pound
model.most_similar(positive=['rupee', 'england'], negative=['india'], topn=5)
```

```python
vec = model['rupee'] - model['india'] + model['england']
model.most_similar(positive=[vec], topn=5)
```
